# Iceberg + MinIO + Spark quickstart

This notebook connects to the `spark-master` cluster using the default `polaris` catalog, backed by MinIO (S3-compatible) storage.

```python
import os
from pyspark.sql import SparkSession

spark = (
    SparkSession.builder
    .appName("iceberg-quickstart")
    .master(os.environ.get("SPARK_MASTER_URL", "spark://spark-master:7077"))
    .getOrCreate()
)
spark
```

```python
spark.sql("CREATE NAMESPACE IF NOT EXISTS nyc")
spark.sql("""
CREATE TABLE IF NOT EXISTS nyc.taxis (
    vendor_id BIGINT,
    trip_date DATE,
    fare_amount DOUBLE
) USING iceberg
""")
spark.sql("INSERT INTO nyc.taxis VALUES (1, DATE '2024-01-01', 12.5), (2, DATE '2024-01-02', 20.0)")
spark.sql("SELECT * FROM nyc.taxis").show()
```
